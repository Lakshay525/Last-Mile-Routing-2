```python
# cell 1: one table for the professor
import pandas as pd, matplotlib.pyplot as plt
a = pd.read_csv("../results/exp01_classical.csv"); b = pd.read_csv("../results/exp02_gnn.csv")
df = pd.concat([a, b[~b.method.isin(a.method.unique())]])
t = df.groupby("method")[["gap_vs_driver", "zone_sw", "sim_to_driver", "secs"]].mean().round(3)
t.sort_values("gap_vs_driver")
```

```python
# cell 2: one chart: cost vs similarity trade-off
fig, ax = plt.subplots(figsize=(6, 5))
ax.scatter(t.gap_vs_driver, t.sim_to_driver)
for m, row in t.iterrows(): ax.annotate(m, (row.gap_vs_driver, row.sim_to_driver))
ax.set_xlabel("cost gap vs driver (lower is better)"); ax.set_ylabel("similarity to driver"); plt.show()
```
