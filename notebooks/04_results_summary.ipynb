```python
import pandas as pd, matplotlib.pyplot as plt
cols = ["gap_vs_driver", "zone_sw", "sim_to_driver", "secs"]
a = pd.read_csv("../results/exp01_classical.csv")   # 207 test routes
b = pd.read_csv("../results/exp02_gnn.csv")         # 62 test routes
t1 = a.groupby("method")[cols].mean().round(3).sort_values("gap_vs_driver")
t2 = b.groupby("method")[cols].mean().round(3).sort_values("gap_vs_driver")
print("Classical methods, 207 test routes"); display(t1)
print("GNN experiment, 62 test routes");     display(t2)
```

```python
fig, ax = plt.subplots(figsize=(6, 5))
ax.scatter(t1.gap_vs_driver, t1.zone_sw)
for m, row in t1.iterrows(): ax.annotate(m, (row.gap_vs_driver, row.zone_sw))
ax.set_xlabel("cost vs driver (lower = shorter driving time)"); ax.set_ylabel("zone switches (lower = tidier)")
ax.set_title("Driving time vs tidiness (207 test routes)"); plt.show()
```
